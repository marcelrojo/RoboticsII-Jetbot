```python
from glob import glob
import torchvision
import torch
```

```python
for path in glob("C:\\Users\\adamt\\Desktop\\RoboticsII-Jetbot\\RoboticsII-Jetbot\\models\\resnet\\*.pth"):
    model = torchvision.models.resnet18(weights=None)
    model.fc = torch.nn.Linear(512, 2)
    model.load_state_dict(torch.load(path))
    example_inputs = (torch.randn(3, 224, 224), torch.randn(2))
    onnx_program = torch.onnx.export(model, example_inputs, dynamo=True)
    onnx_program.optimize()
    onnx_program.save(path.replace("resnet", "resnet_onnx").replace(".pth", ".onnx"))
```
